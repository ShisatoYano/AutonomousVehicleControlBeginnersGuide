## Show the animation in the notebook

Run locally, the simulation opens a matplotlib window. A notebook has no window, so this keeps the animation the simulation creates and shows it as an HTML player instead.

```python
import sys

import matplotlib.animation as animation
import matplotlib.pyplot as plt
from IPython.display import HTML

animations = []
if "original_init" not in globals():  # running this cell again must not wrap the wrapper
    original_init = animation.FuncAnimation.__init__


def keep_animation(self, *args, **kwargs):
    original_init(self, *args, **kwargs)
    animations.append(self)


animation.FuncAnimation.__init__ = keep_animation
plt.show = lambda *args, **kwargs: None
```

## Run the simulation

```python
sys.path.append("src/simulations/path_tracking/pure_pursuit_path_tracking")
import pure_pursuit_path_tracking

pure_pursuit_path_tracking.main()
plt.close("all")
HTML(animations[-1].to_jshtml())
```
